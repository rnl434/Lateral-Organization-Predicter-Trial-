```python
import numpy as np
import matplotlib.pyplot as plt
import subprocess
```

```python
### Helpers
def create_1D_shape(amplitudes, wave_indexes, phases, L, N_points=1000):
    """
    Create a 1D shape based on the given amplitudes, wave numbers, and phases.
    
    Parameters:
    amplitudes (list): List of amplitudes for each wave component.
    wave_indexes (list): List of wave indexes for each wave component.
    phases (list): List of phase shifts for each wave component.
    L (float): The length of the domain.
    x (numpy array): Array of x values where the shape will be evaluated.
    N_points (int): Number of points to evaluate the shape at.
    
    Returns:
    numpy array: The resulting 1D shape evaluated at the given x values.
    """
    x = np.linspace(0, L, N_points)
    shape = np.zeros_like(x)
    wave_numbers = [2 * np.pi * wave_index / L for wave_index in wave_indexes]
    
    
    for amplitude, wave_number, phase in zip(amplitudes, wave_numbers, phases):
        shape += amplitude * np.cos(wave_number * x + phase)
    
    return shape
```

```python




### Controller
amplitudes = [1, 0.5, 0.2]
wave_numbers = [1, 2, 3]
phases = [0, 0, 0]
L = 60
N_points = 1000



shape = create_1D_shape(amplitudes, wave_numbers, phases, L, N_points)
x = np.linspace(0, L, N_points)



fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, shape)
ax.set_xlabel("x (nm)")
ax.set_ylabel("y (nm)")
ax.set_title("1D Shape Created from Multiple Wave Components")
```
